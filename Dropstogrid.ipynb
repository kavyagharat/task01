```python
import os

root = "/content/DropsToGrid"

for dirpath, dirnames, filenames in os.walk(root):
    dirnames[:] = [d for d in dirnames if d != ".git"]

    for filename in filenames:
        if filename.endswith(".py"):
            path = os.path.join(dirpath, filename)

            try:
                with open(path, "r", encoding="utf-8") as f:
                    text = f.read()

                if "FakeDtGDataset" in text:
                    print("\nFOUND:", path)
                    print("=" * 80)

                    # Print the part around FakeDtGDataset
                    index = text.find("FakeDtGDataset")
                    print(text[max(0, index - 1000):index + 5000])

            except Exception:
                pass
```
